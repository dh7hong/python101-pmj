#### Lists
###### 1. Accessing List Elements

```python
fruits = ["apple", "banana", "orange"]

print(f'fruits[0]: {fruits[0]}')
print(f'fruits[1]: {fruits[1]}')
print(f'fruits[2]: {fruits[2]}')
```

```python
fruits = ["apple", "banana", "orange"]

# Negative indexing
print(f'fruits[-1]: {fruits[-1]}')
print(f'fruits[-2]: {fruits[-2]}')
print(f'fruits[-3]: {fruits[-3]}')
```

```python
# List and for loop
fruits = ["apple", "banana", "orange"]

# List and for loop
print("Using for loop to iterate over fruits:")
for i in range(len(fruits)):
    print(f'fruits[{i}]: {fruits[i]}')
```

```python
# changing list items
fruits = ["apple", "banana", "orange"]

fruits[1] = "grape"

print(f'Changed fruits: {fruits}')
```

```python
# adding items
fruits = ["apple", "banana"]

fruits.append("orange")

print(f'Adding fruits: {fruits}')
```

```python
# removing items
fruits = ["apple", "banana", "orange"]
fruits.remove("banana")

print(f'Removed banana: {fruits}')
```

```python
# popping items
fruits = ["apple", "banana", "orange"]
popped_fruit = fruits.pop(1)
print(f'Popped fruit: {popped_fruit}')
print(f'Fruits after popping: {fruits}')
```

```python
# popping items
fruits = ["apple", "banana", "orange"]
popped_minus_one = fruits.pop(-1)
print(f'Popped fruit with -1: {popped_minus_one}')
print(f'Fruits after popping with -1: {fruits}')
```

```python
# popping items
fruits = ["apple", "banana", "orange"]
popped_no_index = fruits.pop()
print(f'Popped fruit with no index: {popped_no_index}')
print(f'Fruits after popping with no index: {fruits}')
```

#### List Slicing and Useful List Functions

```python
# List Slicing
numbers = [10, 20, 30, 40, 50]

print(f'numbers[1:4]: {numbers[1:4]}') # numbers[start:stop]
print(f'numbers[:3]: {numbers[:3]}') # numbers[start:stop]
print(f'numbers[2:]: {numbers[2:]}') # numbers[start:stop]
print(f'numbers[:4:2]: {numbers[:4:2]}') # numbers[start:stop:step]
```

```python
# Function: sorted
numbers = [20, 10, 30, 50, 40]
sorted_numbers = sorted(numbers)
print(f'Sorted numbers: {sorted_numbers}')

reverse_sorted_numbers = sorted(numbers, reverse=True)
print(f'Reverse sorted numbers: {reverse_sorted_numbers}')
```

```python
# Method: sort
numbers = [50, 20, 40, 10, 30]
numbers.sort()
print(f'Sorted numbers: {numbers}')

# Method: reverse
numbers = [50, 20, 40, 10, 30]
numbers.reverse()
print(f'Reversed numbers: {numbers}')
```

#### Tuples

```python
colors = ("red", "green", "blue") # Tuples are like lists but cannot be changed.
print(f'colors[0]: {colors[0]}') # Accessing tuple items is just like lists.

ordinary_list = [1, 2, 3]
ordinary_list[0] = 10 # Lists can be changed.
print(f'ordinary_list: {ordinary_list}')

# but tuples cannot be changed
# colors[0] = "yellow" # This would raise an error

print(type(colors)) # <class 'tuple'>
```

#### Sets

```python
numbers = {1,2,3,4} # Sets store unique items and are unordered.

numbers2 = {1,1,1,2,2,3} # Duplicates are removed automatically.

print(f'numbers set: {numbers}')
print(f'numbers2 set: {numbers2}')
```

```python
# adding items to set
numbers = {1,2,3}
numbers.add(4)
print(f'Added 4: {numbers}')
```

```python
# removing items from set
numbers = {1,2,3}
numbers.remove(2)
print(f'Removed 2: {numbers}')
```

#### Dictionaries

```python
# Dictionaries have key-value pairs and are unordered.
student = {
    "name": "Alice",
    "age": 20,
    "grade": "A"
}
# Dictionary keys, using list() to convert the keys to a list for display
print(f'student keys: {list(student.keys())}')

# Dictionary values, using list() to convert the values to a list for display
print(f'student values: {list(student.values())}')
```

```python
# Accessing dictionary items
student = {
    "name": "Alice",
    "age": 20,
    "grade": "A"
}

print(f'student["name"]: {student["name"]}')

print(f'student["age"]: {student["age"]}')

print(f'student["grade"]: {student["grade"]}')
```

```python
# Adding items to dictionary
student = {
    "name": "Alice",
    "age": 20,
    "grade": "A"
}


student["major"] = "Computer Science"
print(f'Added major: {student}')
```

```python
# Looping through dictionaries
student = {
    "name": "Alice",
    "age": 20
}

for key, value in student.items():
    print(key, value)
```

#### Nested Collections

```python
students = [
    {
        "name": "Alice",
        "grade": "A"
    },
    {
        "name": "Bob",
        "grade": "B"
    }
]

print(f'students: {students}')

print(f'students[0]: {students[0]}')
print(f'students[1]: {students[1]}')

print(f'accessing students[0]["name"]: {students[0]["name"]}')
print(f'accessing students[0]["grade"]: {students[0]["grade"]}')
print(f'accessing students[1]["name"]: {students[1]["name"]}')
print(f'accessing students[1]["grade"]: {students[1]["grade"]}')
```
